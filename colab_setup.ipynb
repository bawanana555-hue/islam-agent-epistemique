## 📦 ÉTAPE 2 : Installez les dépendances

```python
print("📥 Installation des dépendances...")
import subprocess
import sys

dependencies = [
    "fastapi",
    "uvicorn",
    "sentence-transformers",
    "scikit-learn",
    "numpy",
    "google-generativeai",
    "python-multipart"
]

for package in dependencies:
    print(f"  ⏳ Installation de {package}...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package])

print("✅ Toutes les dépendances sont installées!")
```

## 📂 ÉTAPE 3 : Clonez le projet depuis GitHub

## 🚀 ÉTAPE 4 : Lancez le serveur FastAPI (Backend)

```python
import subprocess
import time
import os

os.environ["GEMINI_API_KEY"] = GEMINI_API_KEY

print("🚀 Lancement du serveur FastAPI (Backend)...")
print("⏳ Veuillez patienter 5-10 secondes...\n")

# Lancer uvicorn en arrière-plan
process_api = subprocess.Popen(
    ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

print("✅ Serveur FastAPI lancé sur le port 8000")
print("📍 API disponible sur: http://0.0.0.0:8000")
print("\n⏳ Attente de 3 secondes avant le serveur HTTP...")
time.sleep(3)
```

## 🌐 ÉTAPE 5 : Lancez le serveur HTTP (Frontend)

```python
import subprocess
import time

print("🌐 Lancement du serveur HTTP (Frontend)...")

# Lancer le serveur HTTP
process_http = subprocess.Popen(
    ["python", "-m", "http.server", "8080"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

time.sleep(2)

print("✅ Serveur HTTP lancé sur le port 8080")
print("🌐 Interface disponible sur: http://0.0.0.0:8080")
```

## 🔗 ÉTAPE 6 : Obtenez l'URL publique

```python
!pip install -q pyngrok

from pyngrok import ngrok
import time

print("🔗 Création d'un tunnel public avec ngrok...\n")

# Créer un tunnel pour le serveur HTTP
public_url = ngrok.connect(8080)

print("\n" + "="*70)
print("✅ SUCCÈS ! L'application est maintenant en ligne !")
print("="*70)
print(f"\n🌍 URL PUBLIQUE (accessible depuis Android):")
print(f"\n   {public_url}\n")
print("📱 Ouvrez cette URL dans le navigateur de votre Android:")
print(f"   {public_url}/index.html")
print("\n💡 Conseil: Mettez cette page en favoris!")
print("\n" + "="*70)
```

## 📝 ÉTAPE 7 : Testez l'application

```python
import requests
import json

print("🧪 Test de l'API...\n")

try:
    # Test 1: Vérifier que le serveur API fonctionne
    response = requests.get("http://localhost:8000/")
    if response.status_code == 200:
        print("✅ Serveur API fonctionne correctement")
        print(f"   Réponse: {response.json()}")
    else:
        print(f"❌ Erreur API: {response.status_code}")
    
    # Test 2: Faire une requête de test
    print("\n⏳ Test d'une question (Qu'est-ce que le Tawhid?)...")
    response = requests.get(
        "http://localhost:8000/ask",
        params={"query": "Qu'est-ce que le Tawhid?"}
    )
    
    if response.status_code == 200:
        data = response.json()
        print("✅ Requête API réussie!")
        print(f"\n📝 Question: {data['question']}")
        print(f"\n💬 Réponse:\n{data['answer'][:200]}...")
    else:
        print(f"❌ Erreur: {response.status_code}")
        print(response.text)
        
except Exception as e:
    print(f"❌ Erreur lors du test: {str(e)}")
    print("\n💡 Conseil: Attendez 10 secondes que les serveurs se lancent complètement")
```
